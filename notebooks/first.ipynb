```python
import numpy as np
from tensorflow import keras
from PIL import Image
```

```python
model= keras.models.load_model(r"C:\Users\GIBSON\Desktop\Banana\banana_model.keras")
```

```python
class_names= ["overripe", "ripe", "rotten", "unripe"]
def preprocess_image(image_path):
    img= Image.open(image_path).convert("RGB")
    img= img.resize((224,224))
    img=np.array(img)
    img= np.expand_dims(img, axis = 0)
    predict = model.predict(img)
    then= np.argmax(predict)
    if then == 0:
        print("overripe")
    elif then== 1:
        print("ripe")
    elif then==2:
        print("Rotten")
    else:
        print("unripe")
```
